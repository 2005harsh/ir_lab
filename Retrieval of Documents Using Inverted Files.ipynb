```python
import re
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
# Create file.txt with three documents
with open('file.txt', 'w') as f:
    f.write("This is the first word.\n")
    f.write("This is the second text, Hello! How are you?\n")
    f.write("This is the third, this is it now.")
```

```python
with open('file.txt') as f:
    lines = f.readlines()

print("Documents:")
for i, line in enumerate(lines, start=1):
    print("Document", i, ":", line.strip())
```

```python
stop_words = set(stopwords.words('english'))
```

```python
# Word -> list of document numbers
index = {}

for line_no, line in enumerate(lines, start=1):
    for word in re.findall(r'[a-z]+', line.lower()):
        if word not in stop_words:
            index.setdefault(word, [])

            if line_no not in index[word]:
                index[word].append(line_no)
```

```python
print("Inverted Index:")

for word, line_nos in index.items():
    print(word, "->", line_nos)
```

```python
query = input("Enter a word to search: ").lower()

if query in index:
    print("Found in lines:", index[query])

    for n in index[query]:
        print("Line", n, ":", lines[n - 1].strip())

else:
    print("Word not found")
```
